## 📅 Journal de bord – Jour 13/01/2026

### 🔧 Étapes / Actions réalisées

- Entraînement de 3 types de modèles différents : Régression Linéaire, Arbre de Décision, et Forêt Aléatoire (Random Forest).
- Afin de tester la robustesse et l'éthique, ces modèles ont été entraînés sur **4 scénarios de données différents** :
-   1. **Toutes les données** : Inclut G1, G2 et toutes les variables socio-démographiques.
-   2. **Données éthiques (RGPD)** : Suppression des variables sensibles potentiellement discriminatoires (ex: job des parents, tuteur).
-   3. **Sans G2** : Pour tester la prédiction à mi-semestre.
-   4. **Sans G1 ni G2** : Prédiction purement basée sur le profil socio-démographique (Cold Start).

### 📝 Observations / Difficultés
- Les scénarios incluant G1 et G2 donnent les meilleurs résultats car ces notes sont très corrélées à G3.
- Cependant, la régression linéaire se montre le plus performante sur les données sans G2
- Les deux meilleurs modèles sont l'arbre de décisions et la forêt aléatoire sur le deuxième scénario. Nous sélectionnerons l'arbre de décisions car son comportement est plus intelligible.

### ✅ Prochaines étapes
- Développer une API pour servir le modèle arbre de décisions du scénario 2.
